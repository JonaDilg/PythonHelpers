```python
import numpy as np
from numpy import array as arr

def get_corners_string(RangeX, RangeY):
    safety_dist = 2
    text = "[" + str(RangeX[0]-safety_dist) + "," + str(RangeY[0]-safety_dist) +"],"
    text += "[" + str(RangeX[1]+safety_dist) + "," + str(RangeY[0]-safety_dist) + "],"
    text += "[" + str(RangeX[1]+safety_dist) + "," + str(RangeY[1]+safety_dist) + "],"
    text += "[" + str(RangeX[0]-safety_dist) + "," + str(RangeY[1]+safety_dist) + "]"
    return text 
```
